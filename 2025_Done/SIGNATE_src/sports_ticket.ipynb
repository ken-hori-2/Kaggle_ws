stadiumデータの収容人数(capa)は予測に使えそうなので、stadium情報も結合しましょう。
stadiumにはidが含まれていないので、idを元に結合することはできません。
しかし、スタジアム名(name)が含まれており、これはtmpのスタジアム名(stadium)と共通しているので、スタジアム名を元に横方向に結合できます。

**異なるカラム名をキーにして結合するには**
 - pd.merge(DataFrame1,DataFrame2,left_on="DataFrame1のカラム名",right_on="DataFrame２のカラム名", how='結合方法')

````python
data = pd.read_csv("data.csv")
data_add = pd.read_csv("data_add.csv")
data_new = pd.concat([data, data_add])
condition = pd.read_csv("condition.csv")
condition_add = pd.read_csv("condition_add.csv")
condition_new = pd.concat([condition, condition_add])
tmp = pd.merge(data_new, condition_new, on='id', how='left')
stadium = pd.read_csv('stadium.csv')

# tmpとstadiumを結合して、data_allに代入してください。結合の時に、tmpの行が全て残るように結合してください。
data_all = pd.merge(tmp, stadium, left_on='stadium', right_on='name', how='left')
# tmp, stadium, data_allの列数を表示させて、`tmpの列数+stadiumの列数 = data_allの列数` となっているか確認してください。
print(tmp.shape[1])
print(stadium.shape[1])
print(data_all.shape[1])
# data_allの上から10行を表示させきちんと結合できているかを確認してください。
print(data_all.head(10))

print(tmp.columns)
print(stadium.columns)
````

月を抜き出そう

gamedayから月の部分を抜き出し、新しくmonthカラムを作成しましょう。

月は先頭の2文字に含まれているので、先頭2文字を抜き出す関数を作成し、apply()関数を使って、gamedayカラムの各要素に適用させましょう。

DataFrameの特定のカラムに関数を適用するには

**DataFrame['カラム名'].apply(関数)**
とします。

新しいカラムを作成する場合は、

**DataFrame['作成するカラム名'] = データ**
とします。

````python
# 先頭2文字を抜き出し、int型に変換する関数get_month()を作成してください。
def get_month(x):
    return int(x[:2]) # int(x[0:2])
# 月を表すmonthカラムを作成してください。
data['month'] = data['gameday'].apply(get_month)
# monthカラムが正しく作成されたか確認してください。
print(data['month'].unique())
````

曜日を抜き出そう

次はgamedayから曜日の部分を抜き出して、新しくweekカラムを作成しましょう。

````python
# 曜日部分を抜き出す関数get_week()を作成してください。
def get_week(x):
    return str(x[6:7]) # x[6:7]
    # return str(x[-2:-1]) # 祝/休も入ってしまう
# 曜日を表すweekカラムを作成してください。
data['week'] = data['gameday'].apply(get_week)
# weekカラムが正しく作成されたか確認。
print(data['week'].unique())

# data['gameday'].head()
# data['gameday'].sample(10)
````

節数を抜き出そう

matchカラムも第○節第○日と、節数と日にちが一緒になっているため扱いづらいです。

matchから節数の部分を抜き出して、新しくmatch_numカラムを作成しましょう。

日にちの方は今回は使用しません。
節数は1桁のものと2桁のものがあるので、今までのように特定の場所の文字を抜き出すやり方では上手くいきません。

2文字目を抜き出すと、第1節、第13節はどちらも1になり、13の3の部分が抜き出せません。

かといって2文字目と3文字目の両方を抜き出すと第1節、第13節は1節、13となり、1桁の場合「節」まで含まれてしまいます。

では、どのようにすればいいのでしょうか。

数字部分は必ず「第」と「節」の間に挟まれているので、「第」と「節」が何文字目かを調べ、x[第の場所+1: 節の場所]とすれば間の数字だけ抜き出せます。

特定の文字の場所を調べるにはfind関数というものがありました。

find関数は、ある特定の文字列が含まれていれば、その文字列の左から数えて最初の文字のインデックスを、なければ-1という値を返します

````python
# 節数を抜き出し、int型に変換する関数get_match()を作成してください。
def get_match(x):
    return int(x[x.find('第')+1: x.find('節')])
# 節数を表すmatch_numカラムを作成してください。
data['match_num'] = data['match'].apply(get_match)
# match_numカラムが正しく作成されたか確認。
print(data['match_num'].unique())

# data['match']
````

湿度の修正

humidityに%がついているので、このままでは数値データとては使えません。

humidityから%を取り除き、小数に変換したものに修正しましょう。

(例) 85% → 0.85

````python
# %を取り除き、小数点に変換する関数get_humidity()を作成してください。
def get_humidity(x):
    # return float(x[: x.find('%')])*0.01 # findで%としなくても％は最後尾なので、-1でOK
    tmp = int(x[:-1])
    int_tmp = tmp*0.01
    return int_tmp #int(x[:-1])*0.01

# 湿度を表すhumidityカラムを作成してください。
data['humidity'] = data['humidity'].apply(get_humidity)
# humidityカラムが正しく作成されたか確認。
print(data['humidity'].unique())
````

基本統計量とは
モデルを作成するにあたっては、モデル作成の材料となるデータの特性・特徴を十分に理解することが大切です。その手段の１つとして、データの散らばり具合、つまりデータ分布を確認する方法が挙げられます。

データ分布の特性・特徴を数量的に表す指標として使われるのが基本統計量です。代表的な基本統計量には、最大値・最小値・平均値・中央値・最頻値・標準偏差・四分位数などがあります。

#	統計量	概要 (pickup)

3	平均値	データの代表値を表現する時に利用される。外れ値に影響を受ける

4	中央値	データを小さい順に並べた時にちょうど中央に位置する値。外れ値に影響を受けにくい

````python
import pandas as pd
data = pd.DataFrame([2,4,7,7,10,12,13,17,20,22,30])

print(data.mean())
print(data.median())
print(data.mode())
print(data.max())
````

**相関係数**

相関度合いを数値化したものを相関係数と言います。-1から1の範囲の値をとり、完全な正の相関があるときは1、完全な負の相関があるときは−１となります。相関が無い場合は0となります。

散布図で確認することで大まかな相関関係を捉えることは出来ましたが、具体的に数値で表すことでより厳密な相関度合いを測ることができます。

pandasでは、相関係数はcorr関数を利用することで算出できます。

$$DataFrameが代入された変数.corr()$$

基本的に相関係数は、数値データと数値データとの間の関係性を計測するものなので、corr関数の結果には質的データは含まれません。また相関係数は欠損値を含んでいると本来は算出できないものなのですが、pandasでは自動的に欠損値を含む行を無視して計算してくれます。

tvは〇〇/〇〇/〇〇という形なので、'／'で分割した後の個数を数えれば放送局数が求められます。

文字列を特定の値で分割するにはsplit関数がありました。

これは分割した後のリストを返すので、リストの要素数を数え上げれば良いです。

````python
# '／'で分割した後の要素数を返す関数get_num()を作成してください。
def get_num(x):
    # print(x.split('／')) # split関数は分割した後のリストを返す
    return len(x.split('／'))

# tvカラムから放送局数を表すtv_numカラムを作成してください。
# data['tv_num'] = get_num(data['tv'])
data['tv_num'] = data['tv'].apply(get_num)

# tv_numカラムが正しく作成されたか確認。
print(data['tv_num'].unique())

# data['tv'].unique()

# 放送局数を表すtv_numカラムが作成できたので、放送局数の違いによって観客数がどう異なるのか見ていきましょう。
````

**不快指数の追加**

放送局から放送局数というカラムを新しく作ることができました。
このように既存のデータから新しい特徴量を作れる場合があります。
他にも新しく作れそうな特徴量はあるでしょうか。

気温と湿度は観客数に関係なさそうでした。

では、これらの特徴量は役に立たないデータなのでしょうか。

実は気温と湿度を使って不快指数というものを計算することができます。

不快指数は観客数に影響があるかもしれないので、

不快指数を求め、関係性を見てみましょう。

不快指数は以下の式で計算されます。

$$
不快指数 = 0.81× 気温+0.01× 湿度 ×(0.99 × 気温-14.3)+46.3
$$
DataFrameではカラム同士の演算ができます。

````python
# 空欄を埋めて不快指数を表すdiカラムを作成してください。
data['di'] = 0.81*data['temperature']+0.01*data['humidity']*(0.99*data['temperature']-14.3)+46.3
# 不快指数が求められたか確認するために、diの１行目の値を確認し、小数点以下を四捨五入した値を変数first_diに代入してください。
# first_di = data['di'].astype(int)[0]
first_di = data['di'][0].astype(int)

print(first_di)

# data.columns

# 不快指数カラムを作成することが出来たので、不快指数と観客数の関係を見ていきましょう。
````

## 評価関数の定義

**RMSE**

評価関数は数値を予測するモデルの予測精度を定量的に測るものとして大変重要であり、予測モデルを作成するタスクに応じて適切に設定する必要があります。

今回の予測モデルでは大きな予測誤差を出さないモデルを作成することとし、この目的に合わせて、評価関数としてRMSE（Root Mean Squared Error）を利用することとします。

RMSEとは実測値とモデルの予測値の差を2乗し、その結果をデータの数だけ足し合わせて平均化して、最後に平方根を取った値です。


$$
\text{RMSE} = \sqrt{\frac{1}{n} \sum_{k=1}^{n} (y_{\text{actual},k} - y_{\text{pred},k})^2}
$$


y_actual,kはk番目の予測対象の実績値、y_pred,kはk番目の予測対象予測値を意味し、nは予測対象となっているデータ行数を意味します。

RMSEは誤差を示す尺度である為、評価値が小さいほど良い予測であることを示します。

RMSEの特徴は、複数の予測対象のうち１つでも大きく予測を外すと、評価関数における評価が大きく悪化します。その為、予測したいタスクにおいて予測を大きく外すことが弊害となるような課題においてRMSEを採用することは良い選択と言えます。

````python
# 今回は、RMSE(二乗平均平方根誤差)を学びました。
# RMSEは汎用性が高く、特に頻繁に用いられる評価指標です。定義を忘れないようにしましょう。
# 尚、numpyで一行でRMSEを求める式は下記となります。
# RMSE = np.sqrt(np.mean(np.power(actual-pred,2)))
````

**RMSEの準備**

それでは実際にpythonでRMSEを計算するための準備をしましょう。

scikit-learnライブラリには様々な評価指標が用意されており、ライブラリを利用することで、簡単に評価指標を計算できます。しかし、今回の評価指標であるRMSEはscikit-learnに実装されていません。その代わり、RMSEの平方根を取る前の値であるMSE(Mean Squared Error)は実装されています。

そこで、今回はMSEを利用し、RMSEを算出することにします。MSEを利用する為には、sklearn.metricsからmean_squared_errorをインポートします。

from sklearn.metrics import mean_squared_error as MSE

引数に実績値が代入されている変数と予測値が代入されている変数を与えることで、それらの間のMSEが計算できます。

MSE(実績値, 予測値)

MSEの平方根を取ることでRMSEが求まります。
平方根はnumpyのsqrt関数を利用することで計算ができます。numpyをnpと省略してインポートしている場合には、変数tmpの平方根は下記のように計算します。

np.sqrt(tmp)

実際にRMSEの計算を行うのはモデルの学習を行った後になりますが、ここではRMSEの計算方法についてまず学びましょう。

````python
# ライブラリのインポート
import numpy as np
from sklearn.metrics import mean_squared_error as MSE

actual = [3,4,6,2,4,6,1]
pred = [4,2,6,5,3,2,3]

# MSEの算出
mse = MSE(actual, pred)
print(mse)

# RMSEの算出
rmse = np.sqrt(mse)
print(rmse)

# 今回は、RMSEの概要と計算方法について学びました。
# 実際の計算としては、scikit-learnライブラリに用意されているmean_squared_error()関数を使い、結果の平方根を取ることで求めることができました。
````

**予測精度の評価（評価データ）**

モデルの汎化性能の評価、という意味合いでは本来評価データだけの予測精度の評価で良いのですが、過学習に注意を払う為には、学習及び評価両方に対する予測精度を算出するのが得策です。過学習とは学習データのみの予測に特化し過ぎてしまい、本来当てたい未知のデータへの予測精度が低くなってしまっている現象のことを言いました。過学習が起きているかの見極め方の1つとしては、学習データと評価データそれぞれの予測精度のギャップを見ることです。過学習では学習データへの予測精度は高いが、評価データへの予測精度が著しく低くなる現象の為、両方の評価値を見ることで、過学習が発生していそうかどうかをある程度推測する事ができます。

それでは、学習データに対する予測精度の評価と同様に、評価データへの予測精度を求める為に、RMSEを計算してみましょう。

````python
select_columns = ['id','y','capa','week','home','away','month','stage','tv_num','weather','last_match','forlan']
dummy_data_all = pd.get_dummies(data[select_columns], drop_first=True)
dummy_data = dummy_data_all[dummy_data_all['id'] <= 16084]
dummy_data_target = dummy_data_all[dummy_data_all['id'] >= 16085]
y_1 = dummy_data['y']
dummy_data = dummy_data.drop(columns=['id','y'])


X_train,X_test,y_train,y_test = train_test_split(dummy_data, y_1, random_state = 1234)

lr = LinearRegression()
lr.fit(X_train, y_train)

y_train_pred = lr.predict(X_train)
rmse_train = np.sqrt(MSE(y_train,y_train_pred))
y_test_pred = lr.predict(X_test)
rmse_test = np.sqrt(MSE(y_test,y_test_pred))

# tmp = pd.DataFrame({'係数':lr.coef_}, index=X_train.columns)
# tmp['係数']
# tmp.groupby(tmp['係数']['tv_num'], tmp['係数']['y'])

# -----
# memo
# 回帰係数は回帰直線の傾きを表し、「xが1増えたときにyがどれだけ増減するか」を示しています。
# 例えば、回帰係数が3であれば、xが1増えるとyは3増えるという意味です。
# 重回帰分析の場合も、他の変数の影響を一定としたときに、その変数が1増えるとyがどれだけ変化するかを示します。

# 今回のような重回帰分析（複数の特徴量を使った回帰）では、
# **xは「各特徴量（説明変数）」**を指します。
# たとえば、あなたのデータセットでは
# 	•	`tv_num`（放送局数）
# 	•	`last_match`（最終節かどうか）
# 	•	`capa`（収容人数）
# 	•	`week`（曜日）
# 	•	など
# これら一つ一つが「x」にあたります。
# まとめ
# 	•	x ＝ 各特徴量（説明変数、入力変数）
# 	•	回帰係数 ＝ その特徴量が1増えたとき、目的変数（y）がどれだけ増減するか

# -----

# 回帰係数（各特徴量の重み）
coef_df = pd.DataFrame({'係数': lr.coef_}, index=X_train.columns)
print("回帰係数（特徴量ごとの重み）:")
print(coef_df)

# 切片（intercept）
print("\n切片（intercept）:")
print(lr.intercept_)

# 係数を表示
print(coef_df)

# 具体的な値を変数で抜き出す
tv_num_coef = coef_df.loc['tv_num', '係数'] if 'tv_num' in coef_df.index else None
last_match_coef = coef_df.loc['last_match', '係数'] if 'last_match' in coef_df.index else None

print(f"tv_numの回帰係数: {tv_num_coef}")
print(f"last_matchの回帰係数: {last_match_coef}")

# operation-tips
# last_matchの係数が約3557になっているので、last_matchの試合はそうでない試合よりも3557人多く予測するモデルになっていることが分かります。
````
