# データ分析100本ノック ベストプラクティスガイド

## 1. データの読み込みと基本操作

### 1.1 データの読み込み
```python
import pandas as pd

# CSVファイルの読み込み
sales_df = pd.read_csv('Sales_Data.csv')
member_df = pd.read_csv('Member_Data.csv')

# データの基本確認
print(sales_df.head())  # 上部5行の確認
print(sales_df.info())  # データ型と欠損値の確認
print(sales_df.describe())  # 基本統計量の確認
```

### 1.2 データの結合
```python
# マージ（結合）
merged_df = pd.merge(sales_df, member_df, on='customer_id', how='left')
```

## 2. データの前処理

### 2.1 欠損値の処理
```python
# 欠損値の確認
print(sales_df.isnull().sum())

# 欠損値の補完方法
# 1. 平均値で補完
sales_df['price'].fillna(sales_df['price'].mean(), inplace=True)

# 2. 中央値で補完
sales_df['price'].fillna(sales_df['price'].median(), inplace=True)

# 3. 最頻値で補完
sales_df['category'].fillna(sales_df['category'].mode()[0], inplace=True)
```

### 2.2 データ型の変換
```python
# 日付型への変換
sales_df['date'] = pd.to_datetime(sales_df['date'])

# カテゴリ型への変換
sales_df['category'] = sales_df['category'].astype('category')
```

## 3. データの集計

### 3.1 基本的な集計
```python
# グループ化による集計
grouped = sales_df.groupby('category')['price'].agg(['mean', 'sum', 'count'])

# クロス集計
cross_tab = pd.crosstab(sales_df['category'], sales_df['region'])
```

### 3.2 時系列データの集計
```python
# 日付ごとの集計
daily_sales = sales_df.groupby(sales_df['date'].dt.date)['price'].sum()

# 月次集計
monthly_sales = sales_df.groupby(sales_df['date'].dt.to_period('M'))['price'].sum()
```

## 4. データの可視化

### 4.1 基本的な可視化
```python
import matplotlib.pyplot as plt
import seaborn as sns

# 棒グラフ
plt.figure(figsize=(10, 6))
sns.barplot(x='category', y='price', data=sales_df)
plt.title('Category vs Price')
plt.show()

# 折れ線グラフ（時系列）
plt.figure(figsize=(12, 6))
daily_sales.plot(kind='line')
plt.title('Daily Sales Trend')
plt.show()
```

### 4.2 高度な可視化
```python
# ヒートマップ
plt.figure(figsize=(10, 8))
sns.heatmap(sales_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# 箱ひげ図
plt.figure(figsize=(10, 6))
sns.boxplot(x='category', y='price', data=sales_df)
plt.title('Price Distribution by Category')
plt.show()
```

## 5. 特徴量エンジニアリング

### 5.1 基本的な特徴量作成
```python
# 日付特徴量
sales_df['year'] = sales_df['date'].dt.year
sales_df['month'] = sales_df['date'].dt.month
sales_df['day_of_week'] = sales_df['date'].dt.dayofweek

# 集計特徴量
sales_df['price_per_category'] = sales_df.groupby('category')['price'].transform('mean')
```

### 5.2 高度な特徴量作成
```python
# 移動平均
sales_df['price_ma7'] = sales_df['price'].rolling(window=7).mean()

# 時系列特徴量
sales_df['price_lag1'] = sales_df['price'].shift(1)
```

## 6. データの出力

### 6.1 基本的な出力
```python
# CSVファイルとして出力
sales_df.to_csv('processed_sales.csv', index=False)

# Excelファイルとして出力
sales_df.to_excel('processed_sales.xlsx', index=False)
```

### 6.2 レポート作成
```python
# 基本統計量の出力
print(sales_df.describe().to_string())

# 集計結果の出力
print(grouped.to_string())
```

1. **データの品質確認**
   - 読み込み後は必ず`head()`、`info()`、`describe()`で基本確認
   - 欠損値や異常値の早期発見が重要

2. **データ型の適切な設定**
   - 日付データは`datetime`型に変換

   - 見やすさを考慮したグラフの設定

4. **特徴量エンジニアリング**
   - ドメイン知識を活かした特徴量作成
   - 時系列データの適切な処理
   - 集計特徴量の活用

5. **コードの可読性**
   - 適切な変数名の使用
   - コメントの追加

# データ分析実践ガイド

## Step 1: 環境のセットアップ
```python
# 必要なライブラリのインポート
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# 日本語表示の設定
plt.rcParams['font.family'] = 'Hiragino Sans'
```

## Step 2: データの読み込みと基本確認
```python
# データの読み込み
sales_df = pd.read_csv('Sales_Data.csv')

# データの基本情報確認
print("=== データの基本情報 ===")
print(sales_df.info())

print("\n=== データの先頭5行 ===")
print(sales_df.head())

print("\n=== 基本統計量 ===")
print(sales_df.describe())

print("\n=== 欠損値の確認 ===")
print(sales_df.isnull().sum())
```

## Step 3: データの前処理
```python
# 1. 日付型への変換
sales_df['date'] = pd.to_datetime(sales_df['date'])

# 2. 欠損値の処理
# 数値データの欠損値を平均値で補完
sales_df['price'] = sales_df['price'].fillna(sales_df['price'].mean())

# カテゴリデータの欠損値を最頻値で補完
sales_df['category'] = sales_df['category'].fillna(sales_df['category'].mode()[0])

# 3. 異常値の処理
# 箱ひげ図で異常値を確認
plt.figure(figsize=(10, 6))
sns.boxplot(x=sales_df['price'])
plt.title('価格の分布')
plt.show()

# 異常値の除去（例：3標準偏差以上を外れ値とする）
z_scores = np.abs((sales_df['price'] - sales_df['price'].mean()) / sales_df['price'].std())
sales_df = sales_df[z_scores < 3]
```

## Step 4: 探索的データ分析（EDA）
```python
# 1. 時系列での売上推移
plt.figure(figsize=(12, 6))
daily_sales = sales_df.groupby(sales_df['date'].dt.date)['price'].sum()
daily_sales.plot(kind='line')
plt.title('日次売上推移')
plt.xlabel('日付')
plt.ylabel('売上')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. カテゴリ別の売上分布
plt.figure(figsize=(10, 6))
sns.barplot(x='category', y='price', data=sales_df)
plt.title('カテゴリ別売上')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. 相関分析
plt.figure(figsize=(10, 8))
sns.heatmap(sales_df.corr(), annot=True, cmap='coolwarm')
plt.title('相関行列')
plt.tight_layout()
plt.show()
```

## Step 5: 特徴量エンジニアリング
```python
# 1. 時系列特徴量の作成
sales_df['year'] = sales_df['date'].dt.year
sales_df['month'] = sales_df['date'].dt.month
sales_df['day_of_week'] = sales_df['date'].dt.dayofweek

# 2. 集計特徴量の作成
# カテゴリ別の平均価格
sales_df['category_avg_price'] = sales_df.groupby('category')['price'].transform('mean')

# 3. 移動平均の計算
sales_df['price_ma7'] = sales_df['price'].rolling(window=7).mean()
```

## Step 6: データの集計と分析
```python
# 1. カテゴリ別の集計
category_summary = sales_df.groupby('category').agg({
    'price': ['mean', 'sum', 'count'],
    'quantity': ['mean', 'sum']
}).round(2)

print("=== カテゴリ別集計 ===")
print(category_summary)

# 2. 月次集計
monthly_summary = sales_df.groupby(sales_df['date'].dt.to_period('M')).agg({
    'price': ['mean', 'sum', 'count'],
    'quantity': ['mean', 'sum']
}).round(2)

print("\n=== 月次集計 ===")
print(monthly_summary)
```

## Step 7: 分析結果の可視化
```python
# 1. 月次売上推移
plt.figure(figsize=(12, 6))
monthly_sales = sales_df.groupby(sales_df['date'].dt.to_period('M'))['price'].sum()
monthly_sales.plot(kind='bar')
plt.title('月次売上推移')
plt.xlabel('年月')
plt.ylabel('売上')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. 曜日別の売上分布
plt.figure(figsize=(10, 6))
sns.boxplot(x='day_of_week', y='price', data=sales_df)
plt.title('曜日別の価格分布')
plt.xlabel('曜日')
plt.ylabel('価格')
plt.tight_layout()
plt.show()
```

## Step 8: 結果の出力
```python
# 1. 処理済みデータの保存
sales_df.to_csv('processed_sales.csv', index=False)

# 2. 分析レポートの作成
with open('analysis_report.txt', 'w', encoding='utf-8') as f:
    f.write("=== データ分析レポート ===\n\n")
    f.write("1. データ概要\n")
    f.write(f"総レコード数: {len(sales_df)}\n")
    f.write(f"分析期間: {sales_df['date'].min()} から {sales_df['date'].max()}\n\n")
    
    f.write("2. カテゴリ別集計\n")
    f.write(category_summary.to_string())
    f.write("\n\n")
    
    f.write("3. 月次集計\n")
    f.write(monthly_summary.to_string())
```
